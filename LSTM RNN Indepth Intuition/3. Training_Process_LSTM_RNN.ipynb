## LSTM Training Example: Restaurant Review Sentiment

Consider a text paragraph as input, and the output is a prediction about whether the food in a restaurant is good or bad. The training data consists of reviews for this restaurant. For example, the paragraph reads:

> "Hey, I went to the restaurant and ordered a burger. The burger looked tasty and crispy. But the burger is not good for health. It has a lot of fats and cholesterol. However, this burger was made with whey protein and only vegetables, so it was good."

The output label for this input is 1, indicating the burger is considered good. This forms our training data.

---

### LSTM Memory Components

In LSTM, there are two important components:

- The long-term memory cell, denoted as $c_{t-1}$, representing the previous cell state.
- The short-term memory, represented by the previous hidden state $h_{t-1}$.

At each time step, the input $x_t$ is fed into the network. The LSTM processes this input through three gates:

- Forget gate
- Input gate and candidate memory
- Output gate

These gates control how information flows and is updated in the memory cells.

The operations inside the LSTM include pointwise multiplication and addition to update the cell state and hidden state. The updated cell state $c_t$ and hidden state $h_t$ are computed based on these gate operations and the input vectors.

---

### Input Representation: Embedding Layer

Each word in the input paragraph is first converted into a vector representation. This is done using an embedding layer, such as Word2Vec. The embedding layer transforms words into vectors that capture semantic relationships.

For example, consider a three-dimensional vector space where each dimension corresponds to a feature such as "good", "bad", and "healthy". Each word is mapped to a vector indicating its relationship to these features.

#### Example of Word Vector

Take the word "tasty". Its vector representation might be:

- Good: 0.9
- Bad: 0.0
- Healthy: 0.1

This vector indicates that "tasty" is strongly associated with "good", not associated with "bad", and slightly associated with "healthy".

Each word in the sentence is converted similarly, and these vectors are fed sequentially into the LSTM at each time step.

---

### Processing Sequence in LSTM

As each word vector is input into the LSTM, the forget gate determines what information to retain or discard from the previous cell state. The input gate and candidate memory add new relevant information based on the current input.

For example, when processing the phrase "burger looked tasty and crispy", the input gate adds information about "tasty" and "crispy" to the memory cell because these attributes are relevant.

When the phrase "burger is not good for health" is processed, the forget gate removes some positive information, and the input gate adds negative sentiment related to health.

---

### Dynamic Vector Updates

The vector values change dynamically as the sequence progresses. For instance, after processing "not good for health":

- Good value might decrease from 0.8 to 0.7
- Bad value might increase from 0.1 to 0.4
- Healthy value might decrease to 0.0

These updates reflect the changing sentiment and context captured by the LSTM's memory cells.

---

### Memory Cell Updates

The LSTM updates its long-term and short-term memory cells by selectively forgetting and adding information. The forget gate controls which parts of the previous memory to discard, while the input gate adds new information from the current input.

This mechanism allows the LSTM to maintain relevant context over long sequences, such as tracking sentiment changes in a restaurant review.

---

### Summary of Gate Functions

- Forget Gate: Decides what information to remove from the previous cell state based on the current input and previous hidden state.
- Input Gate and Candidate Memory: Determine what new information to add to the cell state.
- Output Gate: Controls what information from the cell state is output as the hidden state for the current time step.

These gates work together to update the memory cells and produce outputs that capture the sequence's context.

---

### Conclusion

The training of LSTM RNN involves feeding sequential word vectors into the network, where the gates manage memory updates to capture context and sentiment. The forget gate removes irrelevant information, the input gate adds new relevant information, and the output gate produces the current output. This process allows the LSTM to learn complex dependencies in sequential data such as text reviews.
