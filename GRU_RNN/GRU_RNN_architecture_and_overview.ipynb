## Gated Recurrent Unit (GRU): Overview and Motivation

The Gated Recurrent Unit (GRU) is a variant of the LSTM Recurrent Neural Network (RNN), introduced by Cho et al. in 2014. While LSTM RNNs have existed since the 1980s, GRU emerged more recently and offers significant advantages.

---

### LSTM Architecture Overview

- LSTM RNNs have a separate memory cell for long-term memory and a hidden state for short-term memory.
- The architecture includes three gates:
  - Forget gate
  - Input gate
  - Output gate
- These gates control the flow of information. Long-term memory is denoted as $C_t$ and $C_{t-1}$, while short-term memory is $h_t$ and $h_{t-1}$.
- Pointwise operations and activation functions like tanh are used to update these states.

#### Weights and Complexity in LSTM

- Each gate has associated weights: $W_f$ (forget), $W_i$ (input), $W_o$ (output).
- With large input and hidden dimensions, the number of trainable parameters grows rapidly, increasing training time.

---

### Motivation for GRU

- LSTM RNNs solve long-term dependency problems but are complex and slow to train due to many parameters.
- GRU simplifies the architecture by combining long-term and short-term memory into a single hidden state.
- This single cell acts as both long-term and short-term memory.

### GRU Architecture Components

- GRU has two main gates:
  1. **Update gate ($z_t$):** Controls how much of the previous state to keep.
  2. **Reset gate ($r_t$):** Controls how much of the previous state to forget.
- There is also a temporary (candidate) hidden state $\tilde{h}_t$.

#### Gate Computations

- Update gate: $z_t = \sigma(W_z \cdot [h_{t-1}, x_t])$
- Reset gate: $r_t = \sigma(W_r \cdot [h_{t-1}, x_t])$
- $\sigma$ is the sigmoid activation function.

#### Temporary Hidden State

- $\tilde{h}_t = \tanh(W \cdot [r_t \odot h_{t-1}, x_t])$
- $\odot$ denotes element-wise multiplication.

#### Final Hidden State Update

- $h_t = (1 - z_t) \odot h_{t-1} + z_t \odot \tilde{h}_t$
- The update gate $z_t$ balances retention of past information and incorporation of new information.

---

### Functionality of Reset and Update Gates

- **Reset gate ($r_t$):** Determines how much of the previous hidden state to forget.
- **Update gate ($z_t$):** Controls how much of the previous hidden state versus the candidate hidden state to keep.

#### Example of Reset Gate Operation

Given $h_{t-1} = [0.6, 0.5, 0.3, 0.9]$ and $r_t = [0.2, 0.4, 0.8, 0.2]$:
- $0.6 \times 0.2 = 0.12$
- $0.5 \times 0.4 = 0.20$
- $0.3 \times 0.8 = 0.24$
- $0.9 \times 0.2 = 0.18$

---

### Summary of GRU Operations

- Compute update gate $z_t$ and reset gate $r_t$ using sigmoid activation.
- Compute candidate hidden state $\tilde{h}_t$ using tanh activation on reset-modified $h_{t-1}$ and $x_t$.
- Update final hidden state $h_t$ by combining $h_{t-1}$ and $\tilde{h}_t$ weighted by $z_t$.

This coupled gating mechanism simplifies the architecture compared to LSTM by using fewer gates and parameters.

---

### Advantages of GRU

- Fewer trainable parameters compared to LSTM.
- Faster training and optimization.
- Effective handling of long-term dependencies with a combined memory cell.

GRU is a popular choice for sequence modeling tasks where computational efficiency is important.

---

### Key Takeaways

- GRU is a simpler variant of LSTM introduced in 2014 to handle long and short-term dependencies with fewer parameters.
- GRU combines long-term and short-term memory into a single hidden state, reducing complexity.
- The update gate controls how much of the past information to keep, while the reset gate controls how much past information to forget.
- GRU's architecture leads to fewer trainable parameters and faster training compared to LSTM.
