## Types of RNN Problem Statements

RNNs can be categorized based on the input-output sequence relationship:

- **One-to-One:** Single input and single output.
- **One-to-Many:** Single input and multiple outputs.
- **Many-to-One:** Multiple inputs and single output.
- **Many-to-Many:** Multiple inputs and multiple outputs.

**Examples:**
- **One-to-Many:** Image captioning (one image input generates a sequence of words).
- **Many-to-One:** Sentiment analysis (sequence of words input, single sentiment label output).
- **Many-to-Many:** Language translation (sequence in one language input, sequence in another language output).

---

## Motivation for Bidirectional RNNs

Consider the sentence:

> "Krish likes to eat [blank] in Bangalore."

The missing word depends not only on the previous words ("Krish likes to eat") but also on the upcoming words ("in Bangalore"). For example:
- "Krish likes to eat dosa in Bangalore."
- "Krish likes to eat pizza in Paris."

Simple RNNs process sequences only in the forward direction and cannot utilize future context. This limitation motivates the use of bidirectional RNNs.

---

## Architecture of Bidirectional RNNs

Bidirectional RNNs consist of two separate RNNs:
- **Forward RNN:** Processes the sequence from the first word to the last.
- **Backward RNN:** Processes the sequence from the last word to the first.

Each word in the sequence is input to both RNNs. The forward RNN captures past context, while the backward RNN captures future context.

The outputs from both RNNs at each time step are combined (often by concatenation) to form the final output used for prediction. This allows the model to consider both previous and upcoming words.

---

## Visualization of Bidirectional RNN

- **Forward RNN processes:** $X_1 \rightarrow X_2 \rightarrow X_3 \rightarrow X_4$
- **Backward RNN processes:** $X_4 \rightarrow X_3 \rightarrow X_2 \rightarrow X_1$

At time step $t=3$, the output $Y_3$ is computed by combining the hidden states from both forward and backward RNNs, incorporating context from both directions.


---

## Summary

Bidirectional RNNs enhance the capability of standard RNNs by incorporating context from both past and future states. This is especially useful in NLP tasks such as language modeling, where the meaning of a word depends on surrounding words.

By combining outputs from forward and backward passes, bidirectional RNNs provide richer representations and improve prediction performance compared to unidirectional RNNs.

## Equations for Bidirectional RNN

In a Bidirectional RNN, each time step $t$ processes the input $x_t$ in both forward and backward directions.

### Forward RNN

The forward hidden state is computed as:
$$
\overrightarrow{h}_t = f(\overrightarrow{W} x_t + \overrightarrow{U} \overrightarrow{h}_{t-1} + \overrightarrow{b})
$$

### Backward RNN

The backward hidden state is computed as:
$$
\overleftarrow{h}_t = f(\overleftarrow{W} x_t + \overleftarrow{U} \overleftarrow{h}_{t+1} + \overleftarrow{b})
$$

Where:
- $f$ is the activation function (e.g., tanh or ReLU)
- $\overrightarrow{W}, \overrightarrow{U}, \overrightarrow{b}$ are weights and bias for the forward RNN
- $\overleftarrow{W}, \overleftarrow{U}, \overleftarrow{b}$ are weights and bias for the backward RNN

### Combining Forward and Backward States

The output at time $t$ is typically formed by concatenating or summing the forward and backward hidden states:
$$
y_t = g([\overrightarrow{h}_t ; \overleftarrow{h}_t])
$$

or

$$
y_t = g(\overrightarrow{h}_t + \overleftarrow{h}_t)
$$

Where:
- $g$ is an output activation function (e.g., softmax for classification)
- $[\,;\,]$ denotes concatenation

---

**Summary:**  
At each time step, Bidirectional RNNs compute hidden states in both directions and combine them to produce richer context-aware outputs.
