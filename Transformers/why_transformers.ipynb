## LSTM RNN
### Problems
1. LSTM RNNs solve long-term dependency problems but are complex and slow to train due to many parameters.
### Solution:
    - GRU simplifies the architecture by combining long-term and short-term memory into a single hidden state.
    - This single cell acts as both long-term and short-term memory.

## GRU RNN
### Problems
- Simple RNNs suffer from the vanishing gradient problem.
- LSTM and GRU solve this for many-to-one tasks (like sentiment analysis), but struggle when predictions depend on future context.
- Bidirectional RNNs help by processing sequences in both directions.
- For many-to-many tasks (like language translation), we need a more powerful approach: the encoder-decoder architecture.
### Solution: Encoder Decoder
- For indepth explanation refer: [Open Encoder Decoder Notebook](../Encoder_Decoder/encoder_decoder_overview.ipynb)

## Encoder Decoder
### Problems
- For long sentences, the single context vector can't capture all the important information, especially from earlier words.
- As a result, the decoder's performance drops for long sentences (lower BLEU score).
### Solution: Attention Mechanism

## Attention Mechanism
- Despite improvements, attention mechanisms combined with LSTM still process words **sequentially** based on timestamps. This sequential processing prevents **parallel** execution during training, limiting scalability. Even with bidirectional LSTMs, words are processed one at a time, which is inefficient for large datasets.
### Soultion: Transfomers
- Transformers overcome these limitations by eliminating LSTM and RNN components. Instead, they use a self-attention module that allows all words in a sentence to be processed in parallel. 
- This parallel processing supports scalability and efficient training on large datasets. Input embeddings are sent all at once to the encoder, enabling this parallelism.
