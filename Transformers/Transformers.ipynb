## What and Why To Use Transformers

Transformers are a type of deep learning model used in natural language processing that rely on a self-attention mechanism. Self-attention lets the model weigh the importance of each word in a sentence relative to all other words, which is essential for effectively analyzing and processing natural language. Transformers typically follow an encoder–decoder architecture and are applicable to many tasks, such as machine translation.

## Sequence-to-Sequence Tasks

- Machine translation is a canonical sequence-to-sequence (seq2seq) task: both input and output are sequences of words.
- Example: translating English to French is a many-to-many problem (many input words → many output words).
- Sentence length matters: as sentences grow longer, models must still retain accuracy and capture long-range dependencies.

## Transformers and Self-Attention

- Transformers remove LSTM/RNN components and use a self-attention module instead.
- Self-attention allows every word to attend to every other word in the sentence, enabling:
  - Parallel processing of inputs (not sequential),
  - Better scaling to large datasets,
  - Faster training on parallel hardware.
- Input embeddings are fed to the encoder all at once, which is the key to their parallelism.

## Positional Encoding

- Because transformers process tokens simultaneously (no inherent order), positional encoding is used to provide token position information.
- Positional encodings let the model reason about word order and relative positions in a sequence.
- We'll cover positional encoding details and formulas in the next section.

## Scalability and State-of-the-Art Models

- The parallel nature of transformers makes them highly scalable.
- As dataset size and model capacity increase, transformers have produced many SOTA models (e.g., BERT, GPT).
- Transfer learning with pretrained transformer models enables building powerful systems without training from scratch.

## Multi-Modal Applications

- Transformers are not limited to text — they work for multi-modal tasks combining text and images.
- Example: DALL·E and other image-generation systems use transformer-like architectures to map text prompts to images.
- This versatility expands transformer use beyond NLP into broader AI tasks.

## Contextual Embeddings

- Traditional embedding layers assign fixed vectors to words (same vector regardless of context).
- Contextual embeddings vary with sentence context: the same word gets different vectors depending on surrounding words.
- Example: In "My name is Krish and I play cricket," the embedding for "play" should relate to "Krish" and "cricket." Self-attention produces such context-aware representations.

## Summary of Transformer Advantages

- Parallel processing of tokens enables scalability and efficient training.  
- Self-attention provides contextual embeddings and captures token relationships.  
- Transformers overcome limitations of sequential LSTM/RNN encoder–decoder models.  
- They power SOTA models for NLP and multi-modal tasks via transfer learning.

## Key Takeaways

- Transformers are deep learning models built around self-attention, crucial for modern NLP.  
- Unlike LSTM-based encoder–decoders, transformers process all words in parallel, improving scalability.  
- Self-attention creates contextual embeddings, capturing relationships between words in a sentence.  
- Transfer learning with pretrained transformers (BERT, GPT, etc.) enables powerful, practical models across many domains.
